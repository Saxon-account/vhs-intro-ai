## Array Math

```python
x = np.array([[1, 2], [3, 4]], dtype = np.float64)
y = np.array([[5, 6], [7, 8]], dtype = np.float64)

print(x)
print(y)

```

```python
# Elementwise
# add
x + y
# subtract
x - y
# multiply
x * y
# divide
x / y
```

```python
# Elementwise functions
np.sqrt(x)
```

```python
# Dot product
v = np.array([1, 2])
w = np.array([3, 4])

print(np.dot(v, w))
print(v.dot(w))
print(v @ w)
```

```python
x = np.array([[1, 2, 3], [4, 5, 6]])

print(x)
print(np.sum(x, axis=0))
print(np.sum(x, axis=1))
```

```python
# Max, row max
# Min, row min
print(x.max())
print(x.min())
```

## Assignment 1

```python
# Task 1: Calculate Statistics

# Grades for 5 students in 3 subjects
grades = np.array([
[85, 90, 78], # Student 1
[92, 88, 84], # Student 2
[76, 85, 91], # Student 3
[88, 77, 95], # Student 4
[94, 89, 82] # Student 5
])

#1. Compute and print the average grade for each student.

print(np.average(grades, 1))

#2. Compute and print the average grade for each subject.

print(np.average(grades, 0))

#3. Calculate and print the overall average grade across all students and subjects.

print(np.average(grades))
```

```python
# Task 2: Apply Broadcasting
#Adjust Grades with a Bonus Using Broadcasting:
#Suppose each student gets extra points for every subject as a bonus: `[2, 4, 7]`.
#Use broadcasting to add this bonus to each grade and explain how it is used.

print(grades)

bonus = np.array([2, 3, 7])

print(grades + bonus)
```

```python
# Task 3: Plot data
#Create a bar plot of average grades vs students using Matplotlib and label the axes and give a title.

students = np.array(["Bob", "Sally", "Silly", "Not Bob", "Super Bob"])

aveGrades = np.average(grades,1)

print(aveGrades)
print(students)

plt.bar(students, aveGrades)
plt.xlabel('Students')
plt.ylabel('Grades')
plt.show()
```
